### Relplot vs Scatterplot

```python
# relplot vs scatterplot
'''
Function level
relplot() is figure-level (returns a FacetGrid)
scatterplot() is axes-level (returns a Matplotlib Axes)

Plot types
relplot() can create scatter and line plots (kind="scatter" or kind="line")
scatterplot() creates scatter plots only

Faceting / subplots
relplot() easily supports multiple subplots using row and col
scatterplot() does not support faceting by itself (needs Matplotlib or FacetGrid)

Flexibility & control
relplot() is best for high-level, exploratory analysis across categories
scatterplot() is better for fine-grained control over a single plot

Typical use case
Use relplot() for comparing relationships across subsets in one figure
Use scatterplot() for single, customized plots integrated into existing figures
'''
```

### Relplot

```python
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter',hue='sex',style='time',size='size')
```

#### Figure-level functions

```python
# Figure-level functions
# - They return a FacetGrid
# - Handle figure size, layout, and faceting (row, col, hue)
# - You don’t directly control axes with plt.subplot() or ax=
```

## Categorical Distribution Plots

### Boxplot
A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median, third quartile [Q3] and “maximum”). It can tell you about your outliers and what their values are. Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed.

## Regression Plot

In the simplest invocation, both functions draw a scatterplot of two variables, x and y, and then fit the regression model y ~ x and plot the resulting regression line and a 95% confidence interval for that regression.

### relplot

### Implot

```python
sns.lmplot(data=tips,x='total_bill',y='tip',hue='sex')
```

```python
# residplot
sns.residplot(data=tips,x='total_bill',y='tip')
```

A scatter plot shows the raw relationship between two variables. You'll see
- Direction (positive / negative)
- Strength of relationship
- Clusters or group differences

A residual plot shows errors of a regression model, not the raw data.
Mathematically:
    Residual=𝑦_actual−𝑦_predicted

It computes residuals
It plots:
- x-axis: total_bill
- y-axis: residuals (not tip)

## Paigrid vs pairplot

```python
iris = sns.load_dataset('iris')
```

```python
g = sns.PairGrid(data=iris,hue='species')
# g.map
g.map(sns.scatterplot)
```

```python
sns.pairplot(iris,hue='species')
```

```python
# map_diag -> map_offdiag
g = sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.boxplot)
g.map_offdiag(sns.kdeplot)
```

```python
# map_diag -> map_upper -> map_lower
g = sns.PairGrid(data=iris,hue='species')
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```

```python
# vars
g = sns.PairGrid(data=iris,hue='species',vars=['sepal_width','petal_width'])
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```
