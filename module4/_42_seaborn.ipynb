### Relplot vs Scatterplot

```python
# relplot vs scatterplot
'''
Function level
relplot() is figure-level (returns a FacetGrid)
scatterplot() is axes-level (returns a Matplotlib Axes)

Plot types
relplot() can create scatter and line plots (kind="scatter" or kind="line")
scatterplot() creates scatter plots only

Faceting / subplots
relplot() easily supports multiple subplots using row and col
scatterplot() does not support faceting by itself (needs Matplotlib or FacetGrid)

Flexibility & control
relplot() is best for high-level, exploratory analysis across categories
scatterplot() is better for fine-grained control over a single plot

Typical use case
Use relplot() for comparing relationships across subsets in one figure
Use scatterplot() for single, customized plots integrated into existing figures
'''
```

### Relplot

```python
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter',hue='sex',style='time',size='size')
```

## Distribution Plot

```python
# used for univariate analysis
# used to find out the distribution
# Range of the observation
# Central Tendency
# is the data bimodal?
# Are there outliers?

# Plots under distribution plot
# histplot
# kdeplot
# rugplot
```

```python
# figure level -> displot
# axes level -> histplot -> kdeplot -> rugplot
```

#### Figure-level functions

```python
# Figure-level functions
# - They return a FacetGrid
# - Handle figure size, layout, and faceting (row, col, hue)
# - You don’t directly control axes with plt.subplot() or ax=
```

```python
sns.displot(
    data=tips,
    x="total_bill",
    kde=True,
    hue="sex",
    col="day"
)
```

```python
sns.histplot(data=tips, x="total_bill", bins=30, stat="density")
```

#### Axes-level functions

```python
# Draw onto a specific Matplotlib Axes
# - More control over customization
# - Can be embedded in complex Matplotlib layouts
# - Accept an ax= parameter
```

```python
fig, ax = plt.subplots()
sns.histplot(tips["total_bill"], kde=True, ax=ax)
sns.rugplot(tips["total_bill"], ax=ax)
```

```python
fig, ax = plt.subplots()

sns.histplot(
    data=tips,
    x="total_bill",
    hue="sex",
    kde=True,
    bins=20,
    ax=ax
)
```

```python
fig, ax = plt.subplots()

sns.histplot(
    data=tips,
    x="total_bill",
    hue="time",
    kde=True,
    bins=20,
    ax=ax
)
```

### Displot

```python
# It’s also possible to visualize the distribution of a categorical variable using the logic of a histogram. 
# Discrete bins are automatically set for categorical variables

# countplot
sns.displot(data=tips, x='day', kind='hist') 
```

```python
sns.displot(data=tips, x='day', kind='hist', hue='sex', palette=["#070707", "#f00c7e"])
```

### kdeplot

```python
# Rather than using discrete bins, a KDE plot smooths the observations with a Gaussian kernel, producing a continuous density estimate
sns.kdeplot(data=tips,x='total_bill')
```

```python
sns.displot(data=tips,x='total_bill',kind='kde')
```

```python
sns.displot(data=tips,x='total_bill',kind='kde',hue='sex',fill=True,height=10,aspect=2)
```

### Rugplot

```python
# Plot marginal distributions by drawing ticks along the x and y axes.

# This function is intended to complement other plots by showing the location of individual observations in an unobtrusive way.
sns.kdeplot(data=tips,x='total_bill')
sns.rugplot(data=tips,x='total_bill')
```

### Bivariate histogram

```python
# A bivariate histogram bins the data within rectangles that tile the plot 
# and then shows the count of observations within each rectangle with the fill color

# sns.histplot(data=tips, x='total_bill', y='tip')
sns.displot(data=tips, x='total_bill', y='tip',kind='hist')
```

### Bivariate Kdeplot

```python
# a bivariate KDE plot smoothes the (x, y) observations with a 2D Gaussian
sns.kdeplot(data=tips, x='total_bill', y='tip')
```
