```python
health = sns.load_dataset('healthexp')
health
```

```python
plt.plot(health['Country'],health['Spending_USD'])
plt.plot(health['Country'],health['Life_Expectancy'])
```

#### plt labels

```python
plt.title('Title here')
plt.xlabel('X_lable_here')
plt.ylabel('Y_label_here')
```

```python
# style for lines in plot
plt.plot(health['Country'],health['Life_Expectancy'], 
         color='red',                               # canalso be hex code
         linestyle='dashdot',                       # as name suggest
         linewidth=2,                               # width of line
         marker='+',                                # joint for point
         markersize=15,                             # size of point joint
         label='Life Expectancy'
         )


# to label the line 
plt.legend()

# default loc=best
plt.legend(loc='upper left')

# to limit the data 
plt.ylim(75, 80)
# use for outliers present in dataset


# To increase the readability
plt.grid()

# plt.show() to show graph in other editor pycharm, .py file, etc
plt.show()
```
