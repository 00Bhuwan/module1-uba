### colored scatterplot

```python
iris = sns.load_dataset('iris')
iris['species'] = iris['species'].map({
    'setosa': 0,
    'versicolor': 1,
    'virginica': 2
})
iris.columns
```

```python
plt.scatter(iris['sepal_length'],iris['petal_length'],c=iris['species'],cmap='jet',alpha=0.7)
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')
plt.colorbar()
plt.show()
```

#### Define plot size

```python
fig, ax =plt.subplots(figsize=(10,6))
scatter = ax.scatter(iris['petal_width'], iris['sepal_width'], c=iris['species'], cmap='jet', alpha=0.7)
ax.set(ylabel='sepal Width', xlabel='petal Width', title='Petal vs Sepal width')
plt.colorbar(scatter, ax=ax, label='Species')
plt.show()
```

```python
country_total_usd = (health.groupby("Country")['Spending_USD'].sum())
country_avg_life = (health.groupby("Country")['Life_Expectancy'].mean())
country_avg_life
```

```python
colors = ['red', 'blue', 'black', 'pink', 'green', 'orange']
plt.bar(country_total_usd.index, country_avg_life, color=colors, label=colors)
plt.title('Country vs life expectancy')
# plt.legend(title='life expect')
plt.ylim(70, 80)
plt.show()
```

```python
x = np.arange(len(country_total_usd))  # numeric positions
width = 0.35

plt.bar(x - width/2, country_avg_life, width=width, color="red", label="Avg Life Expectancy")
plt.bar(x + width/2, country_total_usd, width=width, color="green", label="Total Spending (USD)")

plt.xticks(x, country_total_usd.index, rotation=45, ha="right")
plt.ylim(70,80)
plt.legend()
plt.tight_layout()
plt.show()
```

#### Bar level using callable

```python
# Bar labels using a callable

fig, ax = plt.subplots()
bar_container = ax.bar(country_avg_life.index, country_total_usd, color=colors)
ax.set(ylabel="Total Usd", xlabel="Country", title='USD for Health')
ax.bar_label(bar_container, fmt='{:,.0f}')
plt.show()
```

```python
fig, ax = plt.subplots()

bottom = np.zeros(len(country_total_usd))

p1 = ax.bar(
    country_total_usd.index,
    country_avg_life,
    width=0.6,
    label="Avg Life Expectancy"
)

bottom += country_avg_life.values

p2 = ax.bar(
    country_total_usd.index,
    country_total_usd,
    width=0.6,
    bottom=bottom,
    label="Total Spending (USD)"
)

ax.bar_label(p1, label_type='center')
ax.bar_label(p2, label_type='center')

ax.set_title("Life Expectancy and Health Spending by Country")
ax.legend()

plt.ylim(70,100)
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
species = ('Adelie', 'Chinstrap', 'Gentoo')
sex_counts = {
    'Male': np.array([73, 34, 61]),
    'Female': np.array([73, 34, 58]),
}
width = 0.6  # the width of the bars: can also be len(x) sequence


fig, ax = plt.subplots()
bottom = np.zeros(3)

for sex, sex_count in sex_counts.items():
    p = ax.bar(species, sex_count, width, label=sex, bottom=bottom)
    bottom += sex_count

    ax.bar_label(p, label_type='center')

ax.set_title('Number of penguins by gender')
ax.legend()

plt.show()
```

## 3D plots

```python
ax = plt.subplot(projection='3d')

ax.scatter(
    iris['sepal_length'],
    iris['petal_length'],
    iris['petal_width'],
    c=iris['species'].astype('category').cat.codes,             # encodes as series
    cmap='viridis'
)

ax.set_title('iris data analysis')

ax.set_xlabel('sepal length')
ax.set_ylabel('petal length')
ax.set_zlabel('species')
```

## 3D surface plots

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx, yy = np.meshgrid(x,y)

z=xx**2 + yy**2

fig = plt.figure(figsize=(12,8))
ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx, yy, z, cmap='viridis')
fig.colorbar(p)

plt.show()
```

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot(projection='3d')

p = ax.plot_surface(xx,yy,z,cmap='viridis')
fig.colorbar(p)

plt.show()
```

```python
x = np.linspace(-10,10,100)
y = np.linspace(-10,10,100)

xx, yy = np.meshgrid(x,y)

z=xx**2 + yy**2

fig = plt.figure(figsize=(12,8))

ax = plt.subplot()

p = ax.contour(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

```python
z = np.sin(xx) + np.cos(yy)

fig = plt.figure(figsize=(12,8))

ax = plt.subplot()

p = ax.contourf(xx,yy,z,cmap='viridis')
fig.colorbar(p)
```

## Heat map

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

plt.figure(figsize=(10, 5))

# Correlation matrix
corr = iris[['sepal_length', 'sepal_width', 
             'petal_length', 'petal_width']].corr()

# Heatmap
sns.heatmap(
    corr,
    annot=True,
    cmap='viridis',
    linewidths=0.5,
    cbar=True
)

plt.title('Iris Feature Correlation Heatmap', fontsize=18)
plt.show()
```
