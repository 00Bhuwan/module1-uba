```python
import pandas as pd
from sqlalchemy import create_engine
import sqlite3
```

```python
# Create an engine.
engine = create_engine("sqlite:///users.db")
```

```python
con = sqlite3.connect('users.db')

data = pd.read_sql_query('Select * from user;', con)
```
